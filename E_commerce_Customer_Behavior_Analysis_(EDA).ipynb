# ***Order Status Distribution***

```python
plt.figure(figsize=(6,4))

sns.countplot(x='OrderStatus', data=df)

plt.title("Order Status Distribution")
plt.xticks(rotation=20)
plt.show()
```

# ***Payment Method***

```python
plt.figure(figsize=(6,4))

sns.countplot(x='PaymentMethod', data=df)

plt.title("Payment Method")
plt.xticks(rotation=30)
plt.show()
```

# ***Monthly Orders***

```python
monthly_orders = df.groupby("MonthName")["OrderID"].count()

monthly_orders = monthly_orders.reindex([
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
])

monthly_orders.plot(kind="bar", figsize=(10,5))

plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

# ***Monthly Revenue***

```python
monthly_revenue = df.groupby("MonthName")["NetRevenue"].sum()

monthly_revenue = monthly_revenue.reindex([
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
])

monthly_revenue.plot(kind="line", marker="o", figsize=(10,5))

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

# ***Top 10 Products***

```python
top_products = df["Product"].value_counts().head(10)

top_products.plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Selling Products")
plt.xlabel("Product")
plt.ylabel("Orders")
plt.xticks(rotation=45)
plt.show()
```

# ***Revenue by Product***

```python
product_revenue = df.groupby("Product")["NetRevenue"].sum().sort_values(ascending=False).head(10)

product_revenue.plot(kind="bar", figsize=(10,5))

plt.title("Top Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

## Quantity **Distribution**

```python
plt.figure(figsize=(6,4))

sns.histplot(df["Quantity"], bins=20)

plt.title("Quantity Distribution")
plt.show()
```

# ***Unit Price Distribution***

```python
plt.figure(figsize=(6,4))

sns.histplot(df["UnitPrice"], bins=20)

plt.title("Unit Price Distribution")
plt.show()
```

# ***Net Revenue Distribution***

```python
plt.figure(figsize=(6,4))

sns.histplot(df["NetRevenue"], bins=20)

plt.title("Net Revenue Distribution")
plt.show()
```

# ***Coupon Usage***

```python
coupon = df["HasCoupon"].value_counts()

plt.figure(figsize=(5,5))

plt.pie(coupon,
        labels=["No Coupon","Coupon Used"],
        autopct="%1.1f%%")

plt.title("Coupon Usage")
plt.show()
```

# ***Delivered vs Cancelled vs Returned***

```python
status = [
    df["IsDelivered"].sum(),
    df["IsCancelled"].sum(),
    df["IsReturned"].sum()
]

labels = ["Delivered","Cancelled","Returned"]

plt.figure(figsize=(6,6))

plt.pie(status,
        labels=labels,
        autopct="%1.1f%%")

plt.title("Order Outcome")
plt.show()
```

# ***Revenue by Payment Method***

```python
plt.figure(figsize=(8,5))

sns.barplot(data=df,
            x="PaymentMethod",
            y="NetRevenue",
            estimator=sum)

plt.title("Revenue by Payment Method")
plt.xticks(rotation=30)
plt.show()
```

# ***Top Customers***

```python
top_customers = df.groupby("CustomerID")["NetRevenue"].sum().sort_values(ascending=False).head(10)

top_customers.plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Customers")
plt.xlabel("Customer")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

# ***Revenue by Day***

```python
day_revenue = df.groupby("DayName")["NetRevenue"].sum()

day_revenue = day_revenue.reindex([
    "Monday","Tuesday","Wednesday",
    "Thursday","Friday","Saturday","Sunday"
])

day_revenue.plot(kind="bar", figsize=(8,5))

plt.title("Revenue by Day")
plt.ylabel("Revenue")
plt.show()
```

# ***Correlation Heatmap***

```python
numeric = df.select_dtypes(include="number")

plt.figure(figsize=(12,8))

sns.heatmap(numeric.corr(),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

# ***Scatter Plot (Quantity vs Revenue)***

```python
plt.figure(figsize=(7,5))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="NetRevenue"
)

plt.title("Quantity vs Revenue")
plt.show()
```

# ***Box Plot***

```python
plt.figure(figsize=(6,4))

sns.boxplot(y=df["NetRevenue"])

plt.title("Net Revenue Boxplot")
plt.show()
```
